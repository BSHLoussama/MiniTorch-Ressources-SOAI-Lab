## 1. Introduction to Data Structures

Data structures are fundamental ways of organizing and storing data in a program. Python provides four main built-in data structures:

| Data Structure | Ordered | Mutable | Duplicates | Syntax |
|----------------|---------|---------|------------|--------|
| **List**       | ✓       | ✓       | ✓          | `[1, 2, 3]` |
| **Tuple**      | ✓       | ✗       | ✓          | `(1, 2, 3)` |
| **Set**        | ✗       | ✓       | ✗          | `{1, 2, 3}` |
| **Dictionary** | ✓*      | ✓       | ✗ (keys)   | `{'a': 1, 'b': 2}` |

*Dictionary maintains insertion order (Python 3.7+)

### Key Concepts:
- **Ordered**: Elements have a defined position and can be accessed by index
- **Mutable**: Elements can be changed after creation
- **Duplicates**: Whether duplicate values are allowed

## 2. Lists - Ordered & Mutable Collections

Lists are the most versatile data structure in Python. They can store items of different types and can be modified after creation.

### 2.1 Creating Lists

````python
# Empty list
empty_list = []
print(f"Empty list: {empty_list}")
print(f"Type: {type(empty_list)}")

# List with integers
numbers = [1, 2, 3, 4, 5]
print(f"\nNumbers: {numbers}")

# List with mixed data types
mixed = [1, "Hello", 3.14, True, None]
print(f"Mixed types: {mixed}")

# Using list() constructor
from_string = list("Python")
print(f"\nFrom string: {from_string}")
````

### 2.2 Accessing List Elements

Lists support **indexing** and **slicing** to access elements.

```
Index:     0    1    2    3    4
List:    ['a', 'b', 'c', 'd', 'e']
Negative: -5   -4   -3   -2   -1
```

````python
fruits = ["apple", "banana", "cherry", "kiwi", "guava"]

# Positive indexing
print(f"First fruit: {fruits[0]}")
print(f"Third fruit: {fruits[2]}")

# Negative indexing (from the end)
print(f"\nLast fruit: {fruits[-1]}")
print(f"Second last: {fruits[-2]}")

# Slicing: list[start:end:step]
print(f"\nFirst 3 fruits: {fruits[0:3]}")
print(f"From index 2 onwards: {fruits[2:]}")
print(f"Every second fruit: {fruits[::2]}")
print(f"Reversed list: {fruits[::-1]}")
````

### 2.3 Modifying Lists

Lists are **mutable**, meaning we can change their contents after creation.

````python
# Changing a single element
fruits = ["apple", "banana", "cherry"]
print(f"Original: {fruits}")

fruits[1] = "watermelon"
print(f"After change: {fruits}")

# Changing multiple elements with slicing
numbers = [1, 2, 3, 4, 5]
numbers[1:4] = [20, 30, 40]
print(f"\nModified numbers: {numbers}")
````

### 2.4 Essential List Methods

Python provides many built-in methods to work with lists efficiently.

````python
# Creating a sample list
my_list = [1, 2, 3]

# append() - Add element to the end
my_list.append(4)
print(f"After append(4): {my_list}")

# insert() - Add element at specific position
my_list.insert(1, 1.5)  # insert 1.5 at index 1
print(f"After insert(1, 1.5): {my_list}")

# extend() - Add multiple elements
my_list.extend([5, 6])
print(f"After extend([5, 6]): {my_list}")

# remove() - Remove first occurrence of value
my_list.remove(1.5)
print(f"After remove(1.5): {my_list}")

# pop() - Remove and return element (default: last)
popped = my_list.pop()
print(f"Popped: {popped}, List: {my_list}")

# clear() - Remove all elements
copy_list = my_list.copy()
copy_list.clear()
print(f"After clear(): {copy_list}")
````

````python
# More useful methods
numbers = [3, 1, 4, 1, 5, 9, 2]

# count() - Count occurrences
print(f"Count of 1: {numbers.count(1)}")

# index() - Find first occurrence
print(f"Index of 4: {numbers.index(4)}")

# sort() - Sort in place
numbers.sort()
print(f"Sorted: {numbers}")

# reverse() - Reverse in place
numbers.reverse()
print(f"Reversed: {numbers}")

# len() - Get length
print(f"Length: {len(numbers)}")
````

### 2.5 Iterating Over Lists

There are multiple ways to loop through lists in Python.

````python
colors = ["red", "green", "blue"]

# Method 1: Simple for loop
print("Method 1: Simple iteration")
for color in colors:
    print(f"  Color: {color}")

# Method 2: With index using enumerate()
print("\nMethod 2: With index")
for index, color in enumerate(colors):
    print(f"  Index {index}: {color}")

# Method 3: With custom start index
print("\nMethod 3: Custom start index")
for index, color in enumerate(colors, start=1):
    print(f"  Position {index}: {color}")
````

### 2.6 List Comprehensions

List comprehensions provide a concise way to create lists.

**Syntax:**
```python
[expression for item in iterable if condition]
```

````python
# Traditional way
squares_traditional = []
for x in range(10):
    squares_traditional.append(x**2)
print(f"Traditional: {squares_traditional}")

# List comprehension way (more Pythonic)
squares = [x**2 for x in range(10)]
print(f"Comprehension: {squares}")

# With condition - only even squares
even_squares = [x**2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {even_squares}")

# Multiple conditions
numbers = [x for x in range(20) if x % 2 == 0 if x % 3 == 0]
print(f"Divisible by 2 and 3: {numbers}")
````

````python
# Advanced: Nested list comprehension
# Create pairs from two lists
colors = ["red", "green"]
objects = ["apple", "car"]

pairs = [[color, obj] for color in colors for obj in objects]
print(f"Color-object pairs: {pairs}")

# With function calls
words = ["hello", "world", "python"]
lengths = [len(word) for word in words]
print(f"Word lengths: {lengths}")
````

### 2.7 Real-World List Examples

#### Example 1: To-Do List Manager

````python
# Create a to-do list
todo = ["Buy groceries", "Clean house", "Pay bills"]

# Add new tasks
todo.append("Schedule meeting")
todo.append("Go for a run")

# Mark task as complete (remove it)
todo.remove("Clean house")

# Check if a task exists
if "Pay bills" in todo:
    print("⚠️ Don't forget to pay the bills!")

# Display remaining tasks
print("\n📝 To-Do List:")
for i, task in enumerate(todo, 1):
    print(f"  {i}. {task}")
````

#### Example 2: Student Grade Analyzer

````python
# Student grades
grades = [85, 92, 78, 90, 88]

# Add a new grade
grades.append(95)

# Calculate statistics
average = sum(grades) / len(grades)
highest = max(grades)
lowest = min(grades)

print(f"📊 Grade Statistics:")
print(f"  Average: {average:.2f}")
print(f"  Highest: {highest}")
print(f"  Lowest: {lowest}")
print(f"  Total grades: {len(grades)}")

# Grades above average
above_avg = [g for g in grades if g > average]
print(f"  Grades above average: {above_avg}")
````

## 3. Tuples - Ordered & Immutable Collections

Tuples are similar to lists, but they are **immutable** (cannot be changed after creation). This makes them ideal for storing data that shouldn't be modified.

### 3.1 Creating Tuples

````python
# Empty tuple
empty = ()
print(f"Empty tuple: {empty}, Type: {type(empty)}")

# Tuple with elements
numbers = (1, 2, 3, 4, 5)
print(f"Numbers: {numbers}")

# Mixed types
mixed = (1, "Hello", 3.14, True)
print(f"Mixed: {mixed}")

# Single element tuple (note the comma!)
single = (42,)  # Without comma, it's just an int
not_tuple = (42)
print(f"Single element tuple: {single}, Type: {type(single)}")
print(f"Not a tuple: {not_tuple}, Type: {type(not_tuple)}")

# Tuple packing (without parentheses)
packed = 1, 2, 3
print(f"Packed: {packed}, Type: {type(packed)}")
````

### 3.2 Accessing Tuple Elements

Tuples support the same indexing and slicing as lists.

````python
coordinates = (10, 20, 30, 40, 50)

# Indexing
print(f"First element: {coordinates[0]}")
print(f"Last element: {coordinates[-1]}")

# Slicing
print(f"First three: {coordinates[:3]}")
print(f"Reversed: {coordinates[::-1]}")
````

### 3.3 Tuple Immutability

Unlike lists, tuples **cannot be modified** after creation.

````python
# Lists are mutable
my_list = [1, 2, 3]
my_list[0] = 100  # ✓ This works
print(f"Modified list: {my_list}")

# Tuples are immutable
my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 100  # ✗ This will raise an error
except TypeError as e:
    print(f"Error: {e}")
````

### 3.4 Tuple Operations and Methods

````python
# Concatenation
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined = tuple1 + tuple2
print(f"Combined: {combined}")

# Repetition
repeated = tuple1 * 3
print(f"Repeated: {repeated}")

# count() - Count occurrences
numbers = (1, 2, 2, 3, 2, 4)
print(f"\nCount of 2: {numbers.count(2)}")

# index() - Find first occurrence
print(f"Index of 3: {numbers.index(3)}")

# Membership testing
print(f"Is 3 in numbers? {3 in numbers}")
print(f"Is 10 in numbers? {10 in numbers}")
````

### 3.5 Tuple Packing and Unpacking

One of the most powerful features of tuples.

````python
# Packing - Create tuple without parentheses
person = "Alice", 25, "Engineer"
print(f"Packed tuple: {person}")

# Unpacking - Extract values into variables
name, age, job = person
print(f"Name: {name}, Age: {age}, Job: {job}")

# Unpacking with * (extended unpacking)
numbers = (1, 2, 3, 4, 5, 6)
first, *middle, last = numbers
print(f"\nFirst: {first}")
print(f"Middle: {middle}")
print(f"Last: {last}")

# Swapping variables using tuples
a, b = 10, 20
print(f"Before swap: a={a}, b={b}")
a, b = b, a  # Swap without temporary variable!
print(f"After swap: a={a}, b={b}")
````

### 3.6 Nested Tuples

````python
# Nested tuple (tuple of tuples)
matrix = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)

# Access nested elements
print(f"First row: {matrix[0]}")
print(f"Element at [1][2]: {matrix[1][2]}")

# Iterate over nested tuple
print("\nMatrix:")
for row in matrix:
    for element in row:
        print(element, end=" ")
    print()  # New line after each row
````

### 3.7 When to Use Tuples vs Lists

**Use Tuples when:**
- Data should not change (immutable)
- Returning multiple values from functions
- Dictionary keys (tuples can be keys, lists cannot)
- Slightly better performance than lists

**Use Lists when:**
- Data needs to be modified
- Need to add/remove elements frequently

````python
# Example: Function returning multiple values
def get_user_info():
    """Returns user information as a tuple"""
    return "Alice", 25, "alice@email.com"

# Unpack the returned tuple
name, age, email = get_user_info()
print(f"User: {name}, Age: {age}, Email: {email}")

# Example: Tuples as dictionary keys
locations = {
    (40.7128, -74.0060): "New York",
    (51.5074, -0.1278): "London",
    (35.6762, 139.6503): "Tokyo"
}
print(f"\nLocation at (40.7128, -74.0060): {locations[(40.7128, -74.0060)]}")
````

## 4. Sets - Unordered & Unique Collections

Sets are unordered collections of **unique** elements. They are useful for:
- Removing duplicates
- Membership testing
- Mathematical set operations (union, intersection, etc.)

### 4.1 Creating Sets

````python
# Empty set (must use set(), not {})
empty = set()
print(f"Empty set: {empty}, Type: {type(empty)}")

# Note: {} creates an empty dictionary, not a set!
not_a_set = {}
print(f"Empty {}: {not_a_set}, Type: {type(not_a_set)}")

# Set with elements
numbers = {1, 2, 3, 4, 5}
print(f"\nNumbers: {numbers}")

# Sets automatically remove duplicates
duplicates = {1, 2, 2, 3, 3, 3, 4}
print(f"Set removes duplicates: {duplicates}")

# Creating set from list
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(my_list)
print(f"From list: {unique_numbers}")
````

### 4.2 Basic Set Operations

````python
fruits = {"apple", "banana", "orange"}

# add() - Add single element
fruits.add("grape")
print(f"After add: {fruits}")

# Adding duplicate has no effect
fruits.add("apple")
print(f"After adding duplicate: {fruits}")

# remove() - Remove element (raises error if not found)
fruits.remove("banana")
print(f"After remove: {fruits}")

# discard() - Remove element (no error if not found)
fruits.discard("kiwi")  # No error even though kiwi is not in set
print(f"After discard (no error): {fruits}")

# pop() - Remove and return arbitrary element
removed = fruits.pop()
print(f"Popped: {removed}, Remaining: {fruits}")
````

### 4.3 Set Mathematical Operations

Sets support mathematical operations like union, intersection, difference, etc.

````python
# Sample sets
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")

# UNION - All elements from both sets
union = set_a.union(set_b)
# Alternative: set_a | set_b
print(f"\nUnion (A ∪ B): {union}")

# INTERSECTION - Common elements
intersection = set_a.intersection(set_b)
# Alternative: set_a & set_b
print(f"Intersection (A ∩ B): {intersection}")

# DIFFERENCE - Elements in A but not in B
diff_a = set_a.difference(set_b)
# Alternative: set_a - set_b
print(f"Difference (A - B): {diff_a}")

diff_b = set_b.difference(set_a)
print(f"Difference (B - A): {diff_b}")

# SYMMETRIC DIFFERENCE - Elements in either set, but not both
sym_diff = set_a.symmetric_difference(set_b)
# Alternative: set_a ^ set_b
print(f"Symmetric Difference (A △ B): {sym_diff}")
````

### 4.4 Set Relationships

````python
# Subset and Superset
set1 = {1, 2, 3, 4, 5}
set2 = {2, 3}
set3 = {1, 2, 3, 4, 5, 6}

# issubset() - Check if all elements are in another set
print(f"Is {set2} a subset of {set1}? {set2.issubset(set1)}")
print(f"Is {set1} a subset of {set2}? {set1.issubset(set2)}")

# issuperset() - Check if contains all elements of another set
print(f"\nIs {set1} a superset of {set2}? {set1.issuperset(set2)}")
print(f"Is {set3} a superset of {set1}? {set3.issuperset(set1)}")

# isdisjoint() - Check if sets have no common elements
set4 = {6, 7, 8}
print(f"\nAre {set2} and {set4} disjoint? {set2.isdisjoint(set4)}")
````

### 4.5 Practical Set Examples

#### Example 1: Removing Duplicates

````python
# List with duplicates
numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
print(f"Original list: {numbers}")
print(f"Length: {len(numbers)}")

# Remove duplicates using set
unique = list(set(numbers))
print(f"\nUnique numbers: {unique}")
print(f"Length: {len(unique)}")
````

#### Example 2: Finding Unique Words

````python
# Count unique words in a sentence
text = "Python is great and Python is powerful and Python is easy to learn"

# Split into words and convert to lowercase
words = text.lower().split()
print(f"All words: {words}")
print(f"Total words: {len(words)}")

# Get unique words
unique_words = set(words)
print(f"\nUnique words: {unique_words}")
print(f"Unique word count: {len(unique_words)}")
````

#### Example 3: Comparing Student Enrollment

````python
# Students enrolled in different courses
python_students = {"Alice", "Bob", "Charlie", "Diana", "Eve"}
java_students = {"Charlie", "Diana", "Frank", "Grace"}

print("Course Analysis:")
print(f"Python students: {python_students}")
print(f"Java students: {java_students}")

# Students in both courses
both = python_students & java_students
print(f"\nStudents in both courses: {both}")

# Students in Python but not Java
python_only = python_students - java_students
print(f"Python only: {python_only}")

# Students in Java but not Python
java_only = java_students - python_students
print(f"Java only: {java_only}")

# All unique students
all_students = python_students | java_students
print(f"\nAll students: {all_students}")
print(f"Total unique students: {len(all_students)}")
````

## 5. Dictionaries - Key-Value Pairs

Dictionaries store data in **key-value pairs**. Keys must be unique and immutable (strings, numbers, tuples), while values can be any type.

### 5.1 Creating Dictionaries

````python
# Empty dictionary
empty = {}
print(f"Empty dict: {empty}, Type: {type(empty)}")

# Dictionary with data
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A",
    "courses": ["Math", "Physics"]
}
print(f"\nStudent: {student}")

# Using dict() constructor
person = dict(name="Bob", age=25, city="New York")
print(f"Person: {person}")

# Note: Keys must be unique (last value wins)
duplicate_keys = {"a": 1, "b": 2, "a": 3}
print(f"\nDuplicate keys: {duplicate_keys}")  # Only last 'a' is kept
````

### 5.2 Accessing Dictionary Elements

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Access using square brackets
print(f"Name: {student['name']}")
print(f"Grade: {student['grade']}")

# Using get() method (safer - no error if key doesn't exist)
print(f"\nAge: {student.get('age')}")
print(f"Major: {student.get('major')}")  # Returns None
print(f"Major: {student.get('major', 'Not specified')}")  # Custom default

# Difference: [] raises KeyError if key doesn't exist
try:
    print(student['major'])  # This will raise KeyError
except KeyError as e:
    print(f"Error accessing non-existent key: {e}")
````

### 5.3 Modifying Dictionaries

````python
student = {"name": "Alice", "age": 20}
print(f"Original: {student}")

# Update existing value
student["age"] = 21
print(f"After age update: {student}")

# Add new key-value pair
student["grade"] = "A"
print(f"After adding grade: {student}")

# Update multiple values at once
student.update({"age": 22, "city": "Paris"})
print(f"After update: {student}")

# Delete key-value pair
del student["city"]
print(f"After deletion: {student}")

# Remove and return value with pop()
grade = student.pop("grade")
print(f"Popped grade: {grade}, Remaining: {student}")
````

### 5.4 Dictionary Methods

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# keys() - Get all keys
keys = student.keys()
print(f"Keys: {list(keys)}")

# values() - Get all values
values = student.values()
print(f"Values: {list(values)}")

# items() - Get key-value pairs as tuples
items = student.items()
print(f"Items: {list(items)}")

# clear() - Remove all items
copy = student.copy()
copy.clear()
print(f"\nAfter clear: {copy}")
print(f"Original unchanged: {student}")
````

### 5.5 Iterating Over Dictionaries

````python
student = {"name": "Alice", "age": 20, "grade": "A"}

# Iterate over keys (default)
print("Keys:")
for key in student:
    print(f"  {key}")

# Iterate over values
print("\nValues:")
for value in student.values():
    print(f"  {value}")

# Iterate over key-value pairs
print("\nKey-Value pairs:")
for key, value in student.items():
    print(f"  {key}: {value}")
````

### 5.6 Nested Dictionaries

````python
# Dictionary of dictionaries
students = {
    "student1": {
        "name": "Alice",
        "age": 20,
        "grades": [85, 90, 92]
    },
    "student2": {
        "name": "Bob",
        "age": 22,
        "grades": [78, 85, 88]
    }
}

# Access nested values
print(f"Student 1 name: {students['student1']['name']}")
print(f"Student 2 grades: {students['student2']['grades']}")

# Iterate over nested dictionary
print("\nAll students:")
for student_id, info in students.items():
    print(f"\n{student_id}:")
    for key, value in info.items():
        print(f"  {key}: {value}")
````

### 5.7 Dictionary Comprehensions

````python
# Basic dictionary comprehension
squares = {x: x**2 for x in range(6)}
print(f"Squares: {squares}")

# With condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(f"Even squares: {even_squares}")

# From two lists
keys = ["name", "age", "city"]
values = ["Alice", 20, "Paris"]
person = {k: v for k, v in zip(keys, values)}
print(f"\nPerson: {person}")

# Transform existing dictionary
prices = {"apple": 0.5, "banana": 0.3, "orange": 0.6}
doubled = {fruit: price * 2 for fruit, price in prices.items()}
print(f"\nOriginal prices: {prices}")
print(f"Doubled prices: {doubled}")
````

### 5.8 Practical Dictionary Examples

#### Example 1: Word Frequency Counter

````python
# Count word frequency in text
text = "python is great python is powerful python is easy"
words = text.split()

# Method 1: Manual counting
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word frequencies:")
for word, count in frequency.items():
    print(f"  {word}: {count}")
````

#### Example 2: Student Grade Manager

````python
# Student records with multiple attributes
students = {
    "Alice": {"age": 20, "grades": [85, 90, 92], "major": "CS"},
    "Bob": {"age": 22, "grades": [78, 85, 88], "major": "Math"},
    "Charlie": {"age": 21, "grades": [92, 95, 98], "major": "CS"}
}

# Calculate average grade for each student
print("Student Averages:")
for name, info in students.items():
    avg = sum(info["grades"]) / len(info["grades"])
    print(f"  {name}: {avg:.2f}")

# Find CS students
cs_students = [name for name, info in students.items() if info["major"] == "CS"]
print(f"\nCS Students: {cs_students}")
````

#### Example 3: Merging Dictionaries

````python
# Two dictionaries to merge
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "d": 4, "e": 5}

print(f"Dict 1: {dict1}")
print(f"Dict 2: {dict2}")

# Method 1: Using ** unpacking (Python 3.5+)
merged = {**dict1, **dict2}
print(f"\nMerged (dict2 values win): {merged}")

# Method 2: Using | operator (Python 3.9+)
merged2 = dict1 | dict2
print(f"Using | operator: {merged2}")

# Method 3: Using update()
merged3 = dict1.copy()
merged3.update(dict2)
print(f"Using update(): {merged3}")
````

## 6. Real-World Applications

### Application 1: Inventory Management System

````python
# Inventory using dictionary
inventory = {
    "apple": {"quantity": 50, "price": 0.5},
    "banana": {"quantity": 30, "price": 0.3},
    "orange": {"quantity": 40, "price": 0.6}
}

def add_stock(item, quantity):
    """Add stock to inventory"""
    if item in inventory:
        inventory[item]["quantity"] += quantity
        print(f"✓ Added {quantity} {item}(s)")
    else:
        print(f"✗ {item} not in inventory")

def sell_item(item, quantity):
    """Sell item from inventory"""
    if item in inventory:
        if inventory[item]["quantity"] >= quantity:
            inventory[item]["quantity"] -= quantity
            total = quantity * inventory[item]["price"]
            print(f"✓ Sold {quantity} {item}(s) for ${total:.2f}")
        else:
            print(f"✗ Not enough {item} in stock")
    else:
        print(f"✗ {item} not found")

# Test the system
print("Initial inventory:")
for item, details in inventory.items():
    print(f"  {item}: {details['quantity']} @ ${details['price']}")

print("\nTransactions:")
sell_item("apple", 10)
add_stock("banana", 20)
sell_item("orange", 50)  # Not enough stock

print("\nFinal inventory:")
for item, details in inventory.items():
    print(f"  {item}: {details['quantity']} @ ${details['price']}")
````

### Application 2: Contact Book

````python
# Contact book using nested dictionary
contacts = {}

def add_contact(name, phone, email):
    """Add a new contact"""
    contacts[name] = {"phone": phone, "email": email}
    print(f"✓ Added {name} to contacts")

def search_contact(name):
    """Search for a contact"""
    if name in contacts:
        info = contacts[name]
        print(f"\n📞 {name}")
        print(f"  Phone: {info['phone']}")
        print(f"  Email: {info['email']}")
    else:
        print(f"✗ {name} not found")

def list_all_contacts():
    """List all contacts"""
    print("\n📖 All Contacts:")
    for name, info in sorted(contacts.items()):
        print(f"  {name}: {info['phone']}")

# Build contact book
add_contact("Alice", "555-1234", "alice@email.com")
add_contact("Bob", "555-5678", "bob@email.com")
add_contact("Charlie", "555-9012", "charlie@email.com")

search_contact("Alice")
search_contact("David")  # Not found
list_all_contacts()
````

### Application 3: Data Processing Pipeline

````python
# Process student data using all data structures
raw_data = [
    ("Alice", "Math", 85),
    ("Bob", "Math", 78),
    ("Alice", "Physics", 92),
    ("Charlie", "Math", 90),
    ("Bob", "Physics", 85)
]

# Step 1: Get unique students (set)
students = {name for name, _, _ in raw_data}
print(f"Unique students: {students}")

# Step 2: Get unique courses (set)
courses = {course for _, course, _ in raw_data}
print(f"Unique courses: {courses}")

# Step 3: Organize by student (dictionary of lists)
student_grades = {}
for name, course, grade in raw_data:
    if name not in student_grades:
        student_grades[name] = []
    student_grades[name].append((course, grade))

# Step 4: Calculate averages
print("\nStudent Averages:")
for student, grades in student_grades.items():
    avg = sum(grade for _, grade in grades) / len(grades)
    print(f"  {student}: {avg:.1f}")
    for course, grade in grades:
        print(f"    {course}: {grade}")
````

## 7. Summary & Best Practices

### Quick Reference Guide

#### When to Use Each Data Structure:

**Lists** `[]`
- ✓ Need ordered collection
- ✓ Need to modify data frequently
- ✓ Duplicates are allowed
- ✓ Need to access by index

**Tuples** `()`
- ✓ Data should not change (immutable)
- ✓ Returning multiple values from function
- ✓ Dictionary keys (tuples can be keys)
- ✓ Slightly faster than lists

**Sets** `{}`
- ✓ Need unique elements only
- ✓ Fast membership testing (`in` operator)
- ✓ Mathematical operations (union, intersection)
- ✓ Removing duplicates

**Dictionaries** `{key: value}`
- ✓ Need to map keys to values
- ✓ Fast lookup by key
- ✓ Data with named fields
- ✓ Counting/grouping items

### Performance Comparison

| Operation | List | Tuple | Set | Dictionary |
|-----------|------|-------|-----|------------|
| Access by index | O(1) | O(1) | N/A | N/A |
| Access by key | N/A | N/A | N/A | O(1) |
| Search | O(n) | O(n) | O(1) | O(1) |
| Insert/Delete | O(n) | N/A | O(1) | O(1) |
| Memory | Medium | Low | High | High |

### Best Practices

1. **Choose the right structure** based on your needs
2. **Use list comprehensions** for concise code
3. **Use `get()` for dictionaries** to avoid KeyError
4. **Use sets** for removing duplicates
5. **Use tuples** for data that shouldn't change
6. **Name variables clearly** to indicate the structure type
7. **Consider memory and speed** for large datasets
